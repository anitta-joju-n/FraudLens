```python
import numpy as np

n_weeks = 13
today = pd.Timestamp("2026-07-27")
week_dates = [today - pd.Timedelta(weeks=i) for i in range(n_weeks-1, -1, -1)]
week_dates
```

```python
def generate_price_history(original_price, discounted_price, n_weeks=13):
    # small random fluctuation around original price for most weeks
    prices = original_price + np.random.normal(0, original_price * 0.02, n_weeks)
    # last week reflects today's actual sale price
    prices[-1] = discounted_price
    return prices
```

```python
sample_prices = generate_price_history(combined_df.loc[0, 'original_price'], combined_df.loc[0, 'discounted_price'])
sample_prices
```

```python
price_history_matrix = np.array([
    generate_price_history(row.original_price, row.discounted_price)
    for row in combined_df.itertuples()
])
price_history_matrix.shape
```

```python
week_labels = [f"week_{i+1}" for i in range(13)]
price_history_df = pd.DataFrame(price_history_matrix, columns=week_labels)
combined_df = pd.concat([combined_df.reset_index(drop=True), price_history_df], axis=1)
combined_df.head()
```

```python
def inject_price_jacking(price_history):
    corrupted = price_history.copy()
    # spike the last 2-3 weeks before the "sale" week (week 13)
    spike_weeks = range(9, 12)  # weeks 10, 11, 12 (0-indexed: 9,10,11)
    inflation_factor = np.random.uniform(1.15, 1.40)  # 15-40% inflated
    for w in spike_weeks:
        corrupted[w] = corrupted[w] * inflation_factor
    return corrupted
```

```python
sample_row = combined_df.loc[0, ['week_1','week_2','week_3','week_4','week_5','week_6','week_7','week_8','week_9','week_10','week_11','week_12','week_13']].values
jacked = inject_price_jacking(sample_row)
jacked
```

```python
np.random.seed(42)
fraud_fraction = 0.18  # ~18% of products get some fraud pattern
n_fraud = int(len(combined_df) * fraud_fraction)

combined_df['is_fraud'] = 0
combined_df['fraud_type'] = 'none'

fraud_indices = np.random.choice(combined_df.index, size=n_fraud, replace=False)
len(fraud_indices)
```

```python
week_cols = ['week_1','week_2','week_3','week_4','week_5','week_6','week_7','week_8','week_9','week_10','week_11','week_12','week_13']

for idx in fraud_indices:
    original = combined_df.loc[idx, week_cols].values.astype(float)
    corrupted = inject_price_jacking(original)
    combined_df.loc[idx, week_cols] = corrupted
    combined_df.loc[idx, 'is_fraud'] = 1
    combined_df.loc[idx, 'fraud_type'] = 'price_jacking'

combined_df['fraud_type'].value_counts()
```

```python
def inject_fake_reference(price_history, actual_original_price):
    corrupted = price_history.copy()
    # weeks 1-12 stay low/stable, close to the discounted price (the "real" historical price)
    stable_price = corrupted[-1] * np.random.uniform(1.02, 1.10)  # just slightly above discount price
    for w in range(12):
        corrupted[w] = stable_price + np.random.normal(0, stable_price * 0.02)
    return corrupted
```

```python
sample_row2 = combined_df.loc[1, week_cols].values.astype(float)
fake_ref = inject_fake_reference(sample_row2, combined_df.loc[1, 'original_price'])
fake_ref
```

```python
np.random.seed(43)
remaining_fraud_pool = combined_df[combined_df['is_fraud'] == 0].index
fraud_indices_2 = np.random.choice(remaining_fraud_pool, size=n_fraud, replace=False)

for idx in fraud_indices_2:
    original = combined_df.loc[idx, week_cols].values.astype(float)
    corrupted = inject_fake_reference(original, combined_df.loc[idx, 'original_price'])
    combined_df.loc[idx, week_cols] = corrupted
    combined_df.loc[idx, 'is_fraud'] = 1
    combined_df.loc[idx, 'fraud_type'] = 'fake_reference'

combined_df['fraud_type'].value_counts()
```

```python
def inject_yo_yo_pricing(price_history):
    corrupted = price_history.copy()
    base = corrupted[-1]  # discounted price as anchor
    for w in range(12):
        # alternate between high and low swings
        if w % 2 == 0:
            corrupted[w] = base * np.random.uniform(1.20, 1.50)  # spike
        else:
            corrupted[w] = base * np.random.uniform(0.95, 1.10)  # dip back down
    return corrupted
```

```python
sample_row3 = combined_df[combined_df['is_fraud'] == 0].iloc[0][week_cols].values.astype(float)
yo_yo = inject_yo_yo_pricing(sample_row3)
yo_yo
```

```python
np.random.seed(44)
remaining_fraud_pool2 = combined_df[combined_df['is_fraud'] == 0].index
fraud_indices_3 = np.random.choice(remaining_fraud_pool2, size=n_fraud, replace=False)

for idx in fraud_indices_3:
    original = combined_df.loc[idx, week_cols].values.astype(float)
    corrupted = inject_yo_yo_pricing(original)
    combined_df.loc[idx, week_cols] = corrupted
    combined_df.loc[idx, 'is_fraud'] = 1
    combined_df.loc[idx, 'fraud_type'] = 'yo_yo_pricing'

combined_df['fraud_type'].value_counts()
```
